```python
import pandas as pd
# Read file
df = pd.read_csv('data/bsm1-rainy-r1.csv')
```

```python
# Only keep the first 23 variable columns after the time column
df_vars = df.iloc[:, 1:24]  # columns 1 to 23 (Python indexing excludes 24)

# Compute Pearson correlation matrix
corr_matrix = df_vars.corr(method='pearson')

# Get top 3 most strongly correlated variables for each variable
top_correlations = {}

for col in corr_matrix.columns:
    sorted_corr = corr_matrix[col].drop(col).abs().sort_values(ascending=False)
    top_vars = sorted_corr.head(3).index
    top_vals = corr_matrix[col][top_vars].values
    top_correlations[col] = list(zip(top_vars, top_vals))

# Print results
for var, top_corrs in top_correlations.items():
    print(f"\nTop 3 correlations for {var}:")
    for name, val in top_corrs:
        print(f"  {name}: {val:.4f}")
```
